```python
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
nltk.download('stopwords')
```

```python
text = """Machine learning is a branch of Artificial Intelligence.
It helps computers learn from data and make predictions.
Machine learning is widely used in modern applications."""

print(text)
```

```python
text = text.lower()

print("Lowercase Text:")
print(text)
```

```python
text = re.sub(r'[^a-zA-Z\s]', '', text)

print("Text after removing special characters:")
print(text)
```

```python
words = text.split()

print("Tokenized Words:")
print(words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("After Stop Word Removal:")
print(filtered_words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("After Stop Word Removal:")
print(filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("After Stemming:")
print(stemmed_words)
```

```python
final_text = ' '.join(stemmed_words)

print("Final Preprocessed Text:")
print(final_text)
```

```python
print("Original Text:")
print(text)

print("\nPreprocessed Text:")
print(final_text)
```
